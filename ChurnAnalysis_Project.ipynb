```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import sqlite3
```

```python

conn = sqlite3.connect('customer_churn.db')

tables = pd.read_sql("SELECT name FROM sqlite_master WHERE type='table';", conn)

for table_name in tables['name']:
    df = pd.read_sql(f'SELECT * FROM "{table_name}"', conn)
    safe_name = table_name.replace(' ', '_').replace('-', '_')
    globals()[f"df_{safe_name}"] = df
    print(f"Created dataframe: df_{safe_name}")

conn.close()

# pulling out the tables from the database (cusomter_churn) and then creating the tables or dataframe using sqlite3
# use pragma command for getting the information for all the tables present in the dataset and their corresponding colums 
```
